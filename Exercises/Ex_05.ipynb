### Exercises 5.2 & 5.3

Define a signal in the form of a sine function with a period of 40 s in the interval $(0 s, 1000 s)$. In a for loop, add normally distributed noise and stack the noisier signals. Compare the result with the original signal. Vary the standard deviation of the random noise and the number of stacked signals.

(a) Standard deviation: 0.1, number: 10, 100, 1000.

(b) Standard deviation: 0.5, number: 10, 100, 1000. 

Additionally, vary the phase of the signal randomly for each run, uniformly distributed in the interval $(0,\pi)$. Interpret the results.

```python
# Import Python libraries 
# -----------------------
#%matplotlib inline
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np
```

```python
def stack(sigman, M, RANDPHASE):
    
    # Define parameters
    T = 40.                  # period 1 [s]
    dt = .1                  # time sampling [s]
    L = 1000.                # length of the time series [s]
    
    # Random phase
    phasemin = 0.      # minimum phase
    phasemax = np.pi   # maximum phase
    phase_avg = (phasemax - phasemin) / 2. # average phase
    phase_sigma = (phasemax - phase_avg)   # phase standard deviation
    
    # Define sine function ...
    omega = 2. * np.pi / T   # compute circular frequency from period
    t = np.arange(0,L+dt,dt) # compute time vector
    x = np.sin(omega*t)      # compute sine wave
        
    # Add normal distributed noise to the time series
    mu = 0.
    
    # Initialize stack
    N = len(x)
    xstack = np.zeros(N)
    for i in range(M):
        
        # Add random phase to signal
        if(RANDPHASE==True):
            phase = np.random.normal(phase_avg, phase_sigma, 1) # create random phase
            x = np.sin(omega*t + phase)    # add random phase on sine wave                
            
        # Add random noise to time series
        xnoise = np.random.normal(mu, sigman, N) # create random numbers
        x1 = x + xnoise # add random noise to time series 
        
        # Stack time series
        xstack = xstack + x1 
        
    # Divide xstack by number of stacks M
    xstack = xstack / M
        
    # Initialize Plots
    plt.figure(figsize=(10,5))
    
    # Plot time series
    plt.plot(t, xstack, 'b')
    plt.xlabel('Time (s)')
    plt.ylabel('f(t) (s)')
    plt.ylim(-1.5,1.5)
    
    # Estimate and print S/N ratio
    SN = np.sqrt(M) / sigman 
    title = r'Stacked Time Series f(t) = Sin(t) ($S/N = \frac{smax\sqrt{M}}{\sigma_n} = $' + str(SN) + ')'
    
    plt.title(title)
```

We can define the noise amplitude in terms of the standard deviation of the normally distributed random numbers $sigman$ and the size of the stack, $M$ ...

```python
interactive_plot = interactive(stack, sigman=(0.1, 0.5, .1), M=(1, 1002, 10), RANDPHASE = False)
output = interactive_plot.children[-1]
output.layout.height = '500px'
interactive_plot
```
