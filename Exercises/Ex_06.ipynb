### Exercise 6.1 

Write the formulas for the recursive calculation of the STA/LTA detector. Use the considerations presented in Section 4.5.2, “Moving Average”.

**Solution**

According to Eq. (4.21), we define the one-sided moving average with a window length $q$ as

\begin{equation}
\hat{m}_i = \frac{1}{q+1} \sum_{j=0}^q x_{i-j} \tag{4.21}
\end{equation}

The moving average for sample $i$ can be calculated recursively from the moving average for sample $i-1$:

\begin{equation}
\hat{m}_{i} = \hat{m}_{i-1} - \frac{1}{q+1} x_{i-q-1} + \frac{1}{q+1} x_{i} \tag{4.22}
\end{equation}

If we set $x_{i-q-1} := \hat{m}_{i-1}$ and define $\alpha := \frac{1}{q+1}$, we obtain

\begin{equation}
\hat{m}_{i} = \hat{m}_{i-1} (1 - \alpha) + \alpha x_{i} \tag{4.23}
\end{equation}

As we will see later when calculating STA, LTA, and kurtosis, this approximation stabilizes the results of the recursive calculation.

Similarly, recursive formulas can be derived for the $\text{STA}$ and $\text{LTA}$ values. $\text{STA}$ and $\text{LTA}$ are defined as:

\begin{equation}
\text{STA}_i = \frac{1}{q_{\text{STA}}+1} \sum_{j=0}^{q_{\text{STA}}} |x_{i-j}| \notag
\end{equation}

\begin{equation}
\text{LTA}_i = \frac{1}{q_{\text{LTA}}+1} \sum_{j=0}^{q_{\text{LTA}}} |x_{i-j}| \notag
\end{equation}

For the recursive calculation, the formula is analogous to that of the moving average:

\begin{equation}
\text{STA}_{i} = \text{STA}_{i-1} - \frac{1}{q_\text{STA}+1} |x_{i-q_\text{STA}-1}| + \frac{1}{q_\text{STA}+1} |x_{i}| \notag
\end{equation}

\begin{equation}
\text{LTA}_{i} = \text{LTA}_{i-1} - \frac{1}{q_\text{LTA}+1} |x_{i-q_\text{LTA}-1}| + \frac{1}{q_\text{LTA}+1} |x_{i}| \notag
\end{equation}

or 

\begin{equation}
\text{STA}_{i} = \text{STA}_{i-1} (1 - \alpha_{\text{STA}}) + \alpha_{\text{STA}} |x_{i}| \notag
\end{equation}

\begin{equation}
\text{LTA}_{i} = \text{LTA}_{i-1} (1 - \alpha_{\text{LTA}}) + \alpha_{\text{LTA}} |x_{i}| \notag
\end{equation}

where $\alpha_{\text{STA}} = \frac{1}{q_\text{STA}+1}$ and $\alpha_{\text{LTA}} = \frac{1}{q_\text{LTA}+1}$.

After STA and LTA have been calculated recursively, the characteristic function is given by:

\begin{equation}
\text{CF}_i = \frac{\text{STA}_{i}}{\text{LTA}_{i}} \notag
\end{equation}

### Exercise 6.2

Write the formulas for the recursive calculation of kurtosis in a sliding window.

**Solution**

According to Eq. (2.40), the central $n$th statistical moment can be estimated as follows:

\begin{equation}
\hat{\alpha}_i^{(n)} = \frac{1}{q_K} \sum_{j=0}^{q_K} (x_{i-j}-\hat{m}_i)^n \notag
\end{equation}

where the moving average must first be calculated.

The recursive calculation yields:

\begin{equation}
\hat{\alpha}_{i}^{(n)} = \hat{\alpha}_{i-1}^{(n)} - \frac{1}{q_K} (x_{i-q_K-1}-\hat{m}_{i-1})^n + \frac{1}{q_K} (x_{i}-\hat{m}_{i})^n \notag
\end{equation}

or, using the approximation $(x_{i-q_K-1}-\hat{m}_{i-1})^n := \hat{\alpha}_{i-1}^{(n)}$ and $ \alpha_K := \frac{1}{q_K}$, we obtain:

\begin{equation}
\hat{\alpha}_{i}^{(n)} = \hat{\alpha}_{i-1}^{(n)} (1 - \alpha_K) + \alpha_K (x_{i}-\hat{m}_{i})^n \notag
\end{equation}

After calculating $\hat{\alpha}_{i}^{(4)}$ and $\hat{\alpha}_{i}^{(2)}$, the kurtosis is given by:

\begin{equation}
K_i = \frac{\hat{\alpha}_{i}^{(4)}}{(\hat{\alpha}_{i}^{(2)})^2} \notag
\end{equation}

### Exercises 6.3 & 6.4

Define a data series consisting of a pulse and random normally distributed noise. Apply the STA/LTA detector to this synthetic data series. Find the optimal parameters for the length of the STA and LTA, respectively.

Compute the characteristic function by estimating the kurtosis in a sliding window. Compare the result with the characteristic function of the STA/LTA detector.

**Solution**

Implementing the recursive algorithms is relatively simple. First, you calculate the corresponding values for the first sample and then apply the recursion. 

It is important not to zero-pad the data with zeros before the start of the time series, because otherwise the beginning of the time series will be detected as a signal. One way to solve this problem is to “zero-pad” the beginning of the time series with a mirrored version of the time series, trimmed to the window length.

Since the calculation of STA/LTA and kurtosis sometimes involves division by small values, you should also implement regularization of the denominator.

```python
def stalta_kurtosis(sn, wsta, wlta, wk, kurtosis, wavelet):
    
    # Define parameters
    dt = 1.                 # time sampling [s]
    L = 800.                # length of the time series [s]

    # Define time array
    t = np.arange(0,L+dt,dt) # compute time vector
    N = len(t)
    
    #  Add spike impulses to time series x ...
    x = np.zeros(N)
    x[N//2] = 1.
    x[N//3] = .75
    x[N//4] = .5
    
    # Convolve source wavelet with spike response
    if(wavelet==True):
        
        wc = 2. * np.pi / 4.
        src = np.sin(wc*t) * np. exp(-1e-1*t)
        x = np.convolve(src,x)
        x = x[:N]
    
    # ... create normal random noise with standard deviation sn ...
    # Here I set a fixed seed to reproduce the same random numbers every time 
    # I change the parameters sn, wsta, wlta, wk, kurtosis, so the time series 
    # does not change. THIS IS A VERY BAD PRACTICE: DON'T DO THIS IF
    # IF YOUR APPLICATION HAS TO RELY ON RANDOM NUMBERS
    np.random.seed(seed=12345656)
    xnoise = np.random.normal(0., sn, N)
    
    # ... add random noise to trend
    x = x + xnoise
    
    # compute window lengths from [s] -> #samples
    q_sta = (int)(wsta//dt)
    q_lta = (int)(wlta//dt)
    q_k = (int)(wk//dt)
    
    # pad beginning of time series with time reversed time series
    x1 = np.concatenate((np.flipud(x[:q_lta]), x), axis=0) 
    N1 = len(x1)
    
    # compute alpha factors 
    alpha_sta = 1. / (q_sta+1)
    alpha_lta = 1. / (q_lta+1)
    alpha_k = 1. / (q_k)
    
    # compute STA and LTA values 
    # --------------------------
    
    # initialize STA and LTA values
    STA = np.zeros(N)
    LTA = np.zeros(N)
    
    # compute first STA and LTA values
    STA[0] = alpha_sta * np.sum(np.abs(x1[0:q_sta]))
    LTA[0] = alpha_lta * np.sum(np.abs(x1[0:q_lta]))
    
    # loop over time series
    j=1
    for i in range(q_lta+1,N1):
        
        STA[j] = STA[j-1] * (1 - alpha_sta) + alpha_sta * np.abs(x1[i])
        LTA[j] = LTA[j-1] * (1 - alpha_lta) + alpha_lta * np.abs(x1[i])
        
        j = j + 1
        
    # Compute characteristic function
    CF = STA/LTA
    
    # normalize characteristic function
    CF = CF / np.max(CF)
    
    if(kurtosis==True):
        
        # compute average, 2nd and 4th moment m, a2, a4
        
        # initialize arrays with zeros
        m = np.zeros(N)
        a2 = np.zeros(N)
        a4 = np.zeros(N)
    
        # arithmetic average m
        # --------------------
        
        # pad beginning of time series with time reversed time series
        x2 = np.concatenate((np.flipud(x[:q_k]), x), axis=0) 
        N2 = len(x2)
        
        # compute first m values
        m[0] = alpha_k * np.sum(x2[0:q_k], axis=0)
        
        # loop over time series
        j = 1
        for i in range(q_k+1,N2):
            
            m[j] = m[j-1] * (1 - alpha_k) + alpha_k * x2[i]
            
            j = j + 1
        
        # moments a2, a4
        # --------------
        
        # compute first a2, a4 values
        a2[0] = alpha_k * np.sum((x2[0:q_k]-m[0])**2)
        a4[0] = alpha_k * np.sum((x2[0:q_k]-m[0])**4)
        
        # loop over time series
        j = 1
        for i in range(q_k+1,N2):
  
            a2[j] = a2[j-1] * (1 - alpha_k) + alpha_k * (x2[i]-m[j])**2
            a4[j] = a4[j-1] * (1 - alpha_k) + alpha_k * (x2[i]-m[j])**4
            
            j = j + 1
    
        # compute Kurtosis
        epsk = 1e-3
        K = a4 / (a2**2+epsk)
        
        # normalize Kurtosis
        K = K / np.max(K) 
    
    # Initialize Plots
    plt.figure(figsize=(20,10))
    
    plt.subplot(311)
    
    # Plot time series
    plt.plot(t, x, 'b', label = r'time series x = xspike + xnoise')
    plt.ylabel('x(t) (s)')
    plt.legend()
    
    title = 'Time series x(t)'
    plt.title(title)
    
    plt.subplot(312)
    
    # Plot CF
    plt.plot(t, CF, 'c', label = r'Characteristic Function CF')
    
    if(kurtosis==True):
        plt.plot(t, K, 'm', label = r'Kurtosis K')
    
    plt.ylabel('CF(t) (s)')
    
    if(kurtosis==True):
        plt.ylabel('CF(t), K(t) (s)')
    
    plt.legend()
    
    title = 'CF'
    
    if(kurtosis==True):
        title = 'CF, K'
        
    plt.title(title)
    

    plt.subplot(313)
    
    # Plot STA, LTA
    plt.plot(t, STA, 'r', label = r'STA')
    plt.plot(t, LTA, 'g', label = r'LTA')
    plt.xlabel('Time (s)')
    plt.ylabel('STA(t), LTA(t) (s)')
    plt.legend()
    
    title = 'STA and LTA'
    plt.title(title)
```

```python
interactive_plot = interactive(stalta_kurtosis, sn=(.01, 1, .01), wsta=(1, 50, 1), wlta=(1, 600, 1), wk=(1, 600, 1), kurtosis=False, wavelet=False)
output = interactive_plot.children[-1]
output.layout.height = '380px'
interactive_plot
```
