# Ambient Noise

Most seismic methods use active sources with a high signal-to-noise ratio to generate images of the subsurface. However, in addition to active methods, ambient noise can also be utilised as a passive seismic source. By correlating and stacking long noise time series from two stations, the Green’s tensor can be determined. Just as with active seismic data, seismic imaging and inversion techniques can be applied to ambient noise data.
